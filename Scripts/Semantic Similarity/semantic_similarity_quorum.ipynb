```python
# ---
import os
import numpy as np
import pandas as pd

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
# ---
```

## Arquivos dos Quoruns

```python
def load_documents(path):
    documents = []
    ids = []

    for file in os.listdir(path):
        if file.endswith(".txt"):
            with open(os.path.join(path, file), "r", encoding="utf-8") as f:
                text = f.read()
                documents.append(text)
                ids.append(file.replace(".txt", ""))

    return ids, documents
```

## Pré-processamento (normalização dos textos)

```python
def preprocess(text):
    text = text.lower()
    text = text.strip()
    return text
```

## Gerar Embeddings

```python
def generate_embeddings(texts, model_name):
    model = SentenceTransformer(model_name)
    embeddings = model.encode(texts)
    return embeddings
```

## Calcular Matriz de Similaridade

```python
def compute_similarity(embeddings):
    sim_matrix = cosine_similarity(embeddings)
    return sim_matrix
```

## Criação de DataFrame

```python
def create_similarity_dataframe(ids, sim_matrix):
    df = pd.DataFrame(sim_matrix, index=ids, columns=ids)
    return df
```

## Salvar Resultados

```python
def save_outputs(model, embeddings, sim_df):
    np.save(f"..//..//Results//Embeddings/embeddings_{model}.npy", embeddings)
    sim_df.to_csv(f"..//..//Results/Similarity/similarity_matrix_{model}.csv")
```
